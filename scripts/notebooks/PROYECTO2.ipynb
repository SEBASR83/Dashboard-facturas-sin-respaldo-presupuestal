```python

# Revisar la clasificación BASE
print("Distribución de BASE:")
display(
    df["BASE"]
    .fillna("VACÍO")
    .astype(str)
    .str.strip()
    .value_counts(dropna=False)
    .rename_axis("Clasificación")
    .reset_index(name="Cantidad")
)

# Revisar los años registrados
columnas_anio = [
    c for c in df.columns
    if "AÑO" in c.upper()
]

for columna in columnas_anio:
    print(f"\nValores más frecuentes: {columna}")
    display(df[columna].value_counts(dropna=False).head(15))

# Revisar valores y tipos de la variable económica
columna_valor = "VALOR DE LA FACTURA"

print("\nTipo de dato del valor de factura:")
print(df[columna_valor].dtype)

print("\nResumen inicial del valor de factura:")
display(df[columna_valor].describe())
```

```python

import pandas as pd

# Inventario inicial de las variables
diccionario = pd.DataFrame({
    "variable": df.columns,
    "tipo_dato": [str(df[c].dtype) for c in df.columns],
    "registros_totales": len(df),
    "valores_no_nulos": [df[c].notna().sum() for c in df.columns],
    "valores_nulos": [df[c].isna().sum() for c in df.columns],
    "porcentaje_nulos": [
        round(df[c].isna().mean() * 100, 2)
        for c in df.columns
    ],
    "valores_distintos": [df[c].nunique(dropna=True) for c in df.columns]
})

display(diccionario)

# Guardar la evidencia para el documento del proyecto
diccionario.to_csv(
    "/content/diccionario_inicial.csv",
    index=False,
    encoding="utf-8-sig"
)

print("Inventario generado:", diccionario.shape)
print("Archivo guardado: /content/diccionario_inicial.csv")
```

```python
import pandas as pd

# Revisar los estados registrados en BASE
print("Distribución de BASE:")
display(
    df["BASE"]
    .value_counts(dropna=False)
    .rename_axis("Estado_BASE")
    .reset_index(name="Cantidad")
)

# Revisar los años de radicación
print("Distribución por año de radicación:")
display(
    df["AÑO RADICACIÓN"]
    .value_counts(dropna=False)
    .sort_index()
    .rename_axis("Año")
    .reset_index(name="Cantidad")
)

# Revisar combinaciones de estado y año
print("Estado BASE por año:")
display(
    pd.crosstab(
        df["AÑO RADICACIÓN"],
        df["BASE"],
        margins=True,
        dropna=False
    )
)
```

```python

import pandas as pd
from IPython.display import display

# 1. Distribución por estado BASE
print("1. Distribución de registros por BASE")
display(
    df["BASE"]
    .value_counts(dropna=False)
    .rename_axis("BASE")
    .reset_index(name="Cantidad")
)

# 2. Cantidad y valor de las facturas por BASE
print("2. Cantidad y valor total por BASE")
resumen_base = (
    df.groupby("BASE", dropna=False)
      .agg(
          cantidad=("VALOR DE LA FACTURA", "size"),
          valor_total=("VALOR DE LA FACTURA", "sum"),
          valor_promedio=("VALOR DE LA FACTURA", "mean"),
          valor_mediano=("VALOR DE LA FACTURA", "median")
      )
      .reset_index()
)

display(resumen_base)

# 3. Distribución temporal por año de radicación
print("3. Registros y valores por año de radicación")
resumen_anual = (
    df.groupby("AÑO RADICACIÓN", dropna=False)
      .agg(
          cantidad=("VALOR DE LA FACTURA", "size"),
          valor_total=("VALOR DE LA FACTURA", "sum")
      )
      .reset_index()
      .sort_values("AÑO RADICACIÓN")
)

display(resumen_anual)

# 4. Revisión de posibles valores atípicos en los años
print("4. Valores de año de radicación encontrados")
display(
    df["AÑO RADICACIÓN"]
    .value_counts(dropna=False)
    .sort_index()
    .rename_axis("Año")
    .reset_index(name="Cantidad")
)

# 5. Resumen general del valor de las facturas
print("5. Estadísticas descriptivas del valor de la factura")
display(
    df["VALOR DE LA FACTURA"]
    .describe()
    .to_frame("Valor de la factura")
)

# 6. Guardar resultados para documentar el proyecto
resumen_base.to_csv(
    "/content/resumen_por_base.csv",
    index=False,
    encoding="utf-8-sig"
)

resumen_anual.to_csv(
    "/content/resumen_anual.csv",
    index=False,
    encoding="utf-8-sig"
)

print("Análisis terminado. Se guardaron dos archivos CSV.")
```

```python
import pandas as pd

df = pd.read_excel("BASE.xlsx", sheet_name="Hoja1")
col = "VALOR DE LA FACTURA"

q1 = df[col].quantile(0.25)
q3 = df[col].quantile(0.75)
iqr = q3 - q1
limite_superior = q3 + 1.5 * iqr

atipicos = df[df[col] > limite_superior]

print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
print("Umbral superior:", limite_superior)
print("Registros por encima del umbral:", len(atipicos))
print("Valor máximo:", df[col].max())

display(
    atipicos[[col, "AÑO RADICACIÓN", "BASE"]]
    .sort_values(col, ascending=False)
    .head(10)
)
```
